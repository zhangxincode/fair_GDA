```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt

def load_and_preprocess_data(file_path):
    # 读取数据文件
    data = pd.read_csv(file_path, header=None, delimiter=' ')

    # 提取标签、敏感属性和特征数据
    labels = data[0]
    sensitive_attr = data[1]  # 第二列作为敏感属性
    features = data.iloc[:, 2:]  # 从第三列开始为特征

    # 根据标签将数据分为两组
    data0 = data[labels == 0]  # 标签为0的行
    data1 = data[labels == 1]  # 标签为1的行

    return data, data0, data1, sensitive_attr, features

def plot_tsne(data, sensitive_attr, features):
    # 标准化特征数据
    scaler = StandardScaler()
    scaled_features = scaler.fit_transform(features)

    # 使用t-SNE降维
    tsne = TSNE(n_components=2, random_state=42)
    tsne_result = tsne.fit_transform(scaled_features)

    # 创建一个DataFrame存储t-SNE结果
    tsne_df = pd.DataFrame(tsne_result, columns=['TSNE1', 'TSNE2'])
    tsne_df['Sensitive Attribute'] = sensitive_attr
    tsne_df['Label'] = data[0]

    # 绘制t-SNE图
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(tsne_df['TSNE1'], tsne_df['TSNE2'], c=tsne_df['Sensitive Attribute'], cmap='coolwarm', marker='o', s=8)
    plt.title('t-SNE Visualization with Sensitive Attribute and Labels')
    plt.xlabel('TSNE1')
    plt.ylabel('TSNE2')
    plt.show()

# 设置文件路径
file_path = './feature_vis/B3_midfeatrue.txt'

# 加载数据并进行预处理
data, data0, data1, sensitive_attr, features = load_and_preprocess_data(file_path)

# 选择要绘制的数据（可以切换 data, data1, 或 data0）
plot_tsne(data0, sensitive_attr, features)
# plot_tsne(data1, sensitive_attr, features)  # 可以选择 data1
# plot_tsne(data0, sensitive_attr, features)  # 或选择 data0
```

```python

# PCA
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
# 读取数据文件
file_path = './feature_vis/GDAB3_midfeatrue.txt'  # 你的数据文件路径
data = pd.read_csv(file_path, header=None, delimiter=' ')

# 提取标签、敏感属性和特征数据
labels = data[0]  # 标签
sensitive_attr = data[1]  # 敏感属性
features = data.iloc[:, 2:]  # 特征

# 标准化特征数据
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

# 使用PCA降维（将数据降到2维）
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_features)

# 创建一个DataFrame存储PCA结果
pca_df = pd.DataFrame(pca_result, columns=['PCA1', 'PCA2'])
pca_df['Sensitive Attribute'] = sensitive_attr
pca_df['Label'] = labels

# 绘制PCA图
plt.figure(figsize=(10, 8))

# 标签的不同形状
markers = ['o', 's', '^', 'D', 'v']  # 不同的形状：圆圈、方形、三角形、菱形、倒三角形
unique_labels = pca_df['Label'].unique()

# 自定义颜色映射
color_map = {0: 'red', 1: 'blue'}  # 示例：给敏感属性值0红色，1蓝色
colors = pca_df['Sensitive Attribute'].map(color_map)  # 根据敏感属性映射颜色

# 为每个标签绘制不同形状的节点
for i, label in enumerate(unique_labels):
    subset = pca_df[pca_df['Label'] == label]
    plt.scatter(subset['PCA1'], subset['PCA2'],
                c=subset['Label'].map(color_map),  # 使用自定义颜色
                marker=markers[i % len(markers)], label=f'Label {label}', s=30)


plt.title('PCA Visualization with Sensitive Attribute (Custom Color) and Labels (Shape)')
plt.xlabel('PCA1')
plt.ylabel('PCA2')

# 显示图例
plt.legend()
plt.show()
#
```
